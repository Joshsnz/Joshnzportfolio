# Agentic Digital Consultant

**A stateful, multi-turn AI agent** designed to act as a digital consultant for Smart-Based (New Zealand AI consultancy). The system intelligently classifies user intent, maintains conversation state, delivers context-aware **pre-scripted video responses**, guides users through a structured consultation form, handles email verification via confirmation codes, and automatically submits qualified leads to the business.

**Period**: September – December 2024

---

### 📋 Project Overview

Built a complete **agentic AI system** that automates the initial client consultation process — from first greeting to lead qualification and handoff. The agent combines natural conversation with a guided form while providing a rich video experience.

**Key Features:**
- Real-time intent classification using GPT-4
- Persistent state management per conversation thread
- Dynamic delivery of **pre-scripted contextual videos**
- 10-stage consultation form with validation
- Email verification flow with confirmation codes
- Automated lead notification to business email
- Comprehensive structured logging for every conversation

---

### 🏗️ System Architecture

**Backend Components:**
- **FastAPI** – High-performance API layer
- **LangChain + OpenAI GPT-4** – Intent classification and reasoning
- **Custom State Machine** – Manages dialog state, form progress, and user details
- **Async Email System** – ProtonMail SMTP with validation and confirmation codes
- **Structured Logging** – Per-thread JSON logs + internal system logs

**Video Response System:**
- Pre-recorded scripted videos mapped to specific conversation states and form stages
- Dynamically serves the correct video based on current intent and progress

**Frontend:** Nuxt 3 + Vue 3 (chat interface + synchronized video player)

---

### 🔄 Core Conversation Flow

1. **Intent Classification**
   - Uses carefully engineered GPT-4 prompts to detect intents such as: `greeting`, `consultation`, `info`, `services`, `goodbye`, `thanks`, `unknown`
   - High tolerance for natural language variations and typos

2. **Stateful Dialog Management**
   - Maintains separate state for each `thread_id` (session)
   - Tracks: `dialog_state`, `form_stage`, `user_details`, `last_active`, `has_greeted`, `is_finalized`, etc.
   - Automatic cleanup of inactive threads after 1 hour

3. **Scripted Video Responses**
   - 15+ pre-scripted videos mapped to conversation states
   - Examples: `welcome.mp4`, `consultation_1.mp4` … `confirmation_code.mp4`, `goodbye.mp4`
   - Videos are triggered based on current dialog state and form progress

4. **Consultation Form Engine**
   - Multi-stage guided form:
     - Business Problem → Domain → Problem Type → Data Format → Timeline → Expected Solution → Additional Details → Name → Email → Confirmation Code
   - Supports cancellation at any time
   - Validates inputs and handles edge cases

5. **Email Verification & Lead Submission**
   - Sends confirmation code to user’s email
   - Validates code before finalizing
   - Sends complete lead details to business email upon successful completion
