# Source-Grounded AI Development Assistant

**A stateful AI coding and repository-intelligence system** designed to help developers understand, inspect, modify, and reason about complex software projects. The system combines LLM-based reasoning with deterministic project tooling, repository indexing, structured retrieval, patch workflows, and evaluation-driven validation to produce reliable, source-grounded development assistance.

**Period**: 2025 – Present

---

### 📋 Project Overview

Built an advanced **agentic development assistant** that operates across an active codebase, helping with architecture review, implementation planning, source-aware Q&A, file comparison, recent-change summaries, patch generation, and project-level reasoning.

The system is designed to avoid generic chatbot behavior by grounding answers in the actual repository state, tracked files, recent edits, and tool-observed evidence. It supports both interactive GUI use and headless evaluation workflows.

**Key Features:**
- Source-grounded answers over large codebases
- Repository-wide architecture and dependency analysis
- Targeted file understanding and responsibility summaries
- Multi-file comparison and boundary analysis
- Recent-change summaries after patches or edits
- Deterministic command handling for scans, plans, and apply flows
- Hybrid retrieval over indexed project files
- Structured telemetry and evaluation logs
- Headless mode for automated testing and regression evaluation
- GUI-ready architecture for interactive developer workflows

### 🏗️ System Architecture

**Core Components:**
- **C++ Application Core** – High-performance desktop/runtime foundation
- **LLM Integration Layer** – Supports remote model providers for reasoning and synthesis
- **Repository Indexing System** – Tracks active source files, metadata, searchable chunks, and project inventory
- **Hybrid Retrieval Engine** – Combines semantic and keyword-style retrieval for source-aware context selection
- **Prompt Assembly System** – Builds carefully scoped prompts from current project state, retrieval results, summaries, and user intent
- **Patch & Planning Workflow** – Supports structured planning, patch proposal, and controlled application flows
- **Telemetry & Logging** – Captures run traces, decisions, evidence, and execution state for debugging and evaluation
- **Headless Evaluation Harness** – Enables repeatable tests against fixed project fixtures and scenario datasets

**User Interfaces:**
- Interactive GUI for developer-facing usage
- Headless runtime for automated tests, benchmarks, and regression analysis

### 🔄 Core Workflow

1. **Project Understanding**
   - Scans and indexes the active repository
   - Tracks source files, file inventory, project structure, and recent changes
   - Builds searchable context for source-grounded responses

2. **Intent-Aware Request Handling**
   - Differentiates between general chat, codebase questions, architecture reviews, file-specific questions, comparisons, and patch-related requests
   - Handles direct questions such as:
     - “Where is this file used?”
     - “What changed in the project?”
     - “Compare these two systems”
     - “Explain the startup flow”
     - “What file is responsible for this behavior?”

3. **Source-Grounded Retrieval**
   - Retrieves relevant files and code sections from the current repository
   - Supports broad architecture questions as well as narrow file-targeted questions
   - Prioritizes current source evidence over stale summaries or assumptions

4. **Answer Generation**
   - Produces structured, grounded answers based on retrieved project evidence
   - Supports architecture explanations, responsibility summaries, usage traces, recent-update summaries, and code comparisons
   - Avoids unsupported claims when the repository does not provide enough evidence

5. **Patch Planning & Application**
   - Can generate implementation plans and controlled patch workflows
   - Supports deterministic command paths for project scans, planning, and apply actions
   - Tracks post-change state so follow-up questions can refer to recent edits

6. **Evaluation & Regression Testing**
   - Uses a Python-based evaluation harness to run repeatable scenarios
   - Tests retrieval accuracy, answer behavior, patch-follow-up behavior, and source-grounding reliability
   - Produces structured logs for debugging failures and improving system behavior
